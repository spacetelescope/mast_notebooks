## Introduction

This tutorial describes how to load non-Roman source catalogs into `mast-aladin`, allowing you to vizualize the data against a sky background. We will use the [`astroquery.mast`](https://astroquery.readthedocs.io/en/stable/mast/mast.html) library to query for source catalogs from the [PanSTARRS](https://www2.ifa.hawaii.edu/research/Pan-STARRS.shtml) and [James Webb Space Telescope (JWST)](https://science.nasa.gov/mission/webb/) missions. The examples provided will focus on sources near the [Andromeda Galaxy (M31)](https://science.nasa.gov/mission/hubble/science/explore-the-night-sky/hubble-messier-catalog/messier-31/).


## Imports

We will import the following in this tutorial:

- `astroquery.mast`: query and download source catalogs for a subset of the astronomical catalogs stored at MAST
- `mast_aladin.MastAladin`: interactive sky map widget for visualizing source catalogs
- `sidecar`: organize our `mast-aladin` viewer widget

```python
from astroquery.mast import Catalogs, MastMissions
from mast_aladin import MastAladin
from sidecar import Sidecar
```

## Overlay non-Roman Source Catalogs

### Setup
The first step to working with `mast-aladin` is to set up the notebook workspace. We start by initializing an instance of the `mast-aladin` class with the `target` set to `M31` and the `fov`, which controls the zoom level, set to `10` degrees to ensure we can see the source catalogs that we will load later in the notebook. We display the widget in the main panel to the right of the notebook by using [`sidecar`](https://github.com/jupyter-widgets/jupyterlab-sidecar) with the anchor set to `split-right`. More information on this can be found in the [interactive_data_exploration/how_to/organize_widgets/organize_widgets.ipynb](../organize_widgets/organize_widgets.ipynb) tutorial.


```python
mast_aladin = MastAladin(target="M31", fov=10)

sc = Sidecar(title="mast-aladin", anchor="split-right")
# Display the widget within the sidecar
with sc:
    display(mast_aladin)
```

### Query for non-Roman Source Catalogs
Next, we query for non-Roman source catalogs. The [`astroquery.mast.Catalogs`](https://astroquery.readthedocs.io/en/stable/mast/mast_catalog.html) and [`astroquery.mast.MastMissions`](https://astroquery.readthedocs.io/en/latest/mast/mast_missions.html) classes provide access to a subset of the astronomical catalogs stored at MAST. For this example, we will look at the PanSTARRS DR2 and JWST source catalogs.

**Note:** Both classes provide the same query API interface.

**PanSTARRS Source Catalogs**

PanSTARRS has a large number of sources near M31, so we will set a smaller radius of `.02` degrees to quickly retrieve an example `astropy.table.Table`

```python
panstarrs_catalog = Catalogs.query_criteria(
    collection="ps1_dr2",
    object_name="M31",
    radius=0.02
)
```

**JWST Source Catalogs**

JWST has no observations near the center of `M31`, so the radius needs to be set to `30` arcminutes to ensure that results are returned.

```python
mm = MastMissions(mission="jwst")
jwst_observations = mm.query_object(object_names="m31", radius=30)
```

### Overlaying a Source Catalog
You can overlay any `astropy.table.Table` with fields containing `ra` and `dec` values onto the sky viewer using the `add_table()` method. The method returns an `Overlay` object which can be used to modify its styling properties.


```python
panstarrs_overlay = mast_aladin.add_table(panstarrs_catalog, name="panstarrs", raField="raMean", decField="decMean")
jwst_overlay = mast_aladin.add_table(jwst_observations, name="jwst", raField="targ_ra", decField="targ_dec")
```

![alt text](overlaying-a-source-catalog.png)

### Update a Table's Style
You can update the name, color, shape, and source size of an `Overlay`.

```python
# returns the updated Overlay
panstarrs_overlay = panstarrs_overlay.update(
    name="panstarrs-updated",
    color="Blue", 
    shape="circle", 
    source_size=30
)
```

### Removing a Source Catalog
The `remove_overlay` method allows for the removal of source catalogs. 

First, we can view all of the currently loaded `Overlays` with `mast_aladin.overlays`, taking note of the source catalog we wish to remove.

```python
mast_aladin.overlays
```

Next, provide the `name` attached to the source catalog to the `remove_overlay` method.

```python
mast_aladin.remove_overlay('panstarrs-updated')
```
