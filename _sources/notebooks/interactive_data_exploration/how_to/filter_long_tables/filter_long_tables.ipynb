## Introduction
This tutorial demonstrates how to use [`mast-table`](https://github.com/spacetelescope/mast-table) filtering functionality to easily refine query search results to a smaller subset of your choosing. `mast-table` is an interactive table widget for displaying, exploring, and filtering tabular data, designed to support [Mikulski Archive for Space Telescopes (MAST)](https://archive.stsci.edu/) query results. We use [astroquery](https://github.com/astropy/astroquery) to retrieve data from MAST and then vizualize and filter the results with `mast-table`. For more information about using `MastMissions`, see the notebook [`missions_mast_search.ipynb`](https://github.com/spacetelescope/mast_notebooks/blob/main/notebooks/multi_mission/missions_mast_search/missions_mast_search.ipynb).

## Imports
We will leverage the `mast-table` package to display and filter data, and we will use `astroquery` to populate the table widgets with example data.

- `mast-table` to create filterable table widgets
- `astroquery` for retrieving example data to view

```python
from mast_table import MastTable
from astroquery.mast import MastMissions
```

## Filtering Long Tables with `mast-table`

### Retrieving MAST Data

First, we will get example MAST data from `astroquery` using the cell below. We are retrieving MAST data on [Messier 1](https://science.nasa.gov/mission/hubble/science/explore-the-night-sky/hubble-messier-catalog/messier-1/) from the [James Webb Space Telescope (JWST)](https://www.stsci.edu/jwst). By running the cell below, we can see the astroquery `results` table.

```python
# Create MastMissions object to search for JWST datasets
missions = MastMissions(mission='jwst')

# Query for Messier 1 ('M1')
results = missions.query_object('M1')

# Display the length of the astroquery results to compare
print("table length:", len(results))
```

### Creating a `mast-table`

We will now create an instance of `mast-table` using the cell below. We are displaying the widget in a sidecar in the bottom half of the screen.

```python
# Create a mast-table instance
mast_table = MastTable(results)

# Display the widget in sidecar split bottom
mast_table.show("sidecar:split-bottom")
```

Now, looking at the table footer, we see the length of the `mast-table` widget matches the length of the astroquery results table from the cell above.

### Sorting `mast-table` Columns

Take a moment to look through the columns available in the table. 

Let's say that you are specifically interested in knowing when the first observations of M1 were taken by JWST. Using the `date_obs` column, hover over the column name and you will see a sorting arrow appear, like the one below:

<img src="arrow_up.png" width="100">

Clicking once will sort the column descendingly, clicking a second time will sort the column ascendingly, and clicking for a third time will remove any applied sorting. 

Using column sorting, identify the earliest `date_obs` of M1 by JWST.

### Applying Filters in `mast-table`

Consider the scenario you are now only interested in a specific subset of the data: mid-infrared data with small angular separation. 

Instead of needing to run a new query or filter the astroquery results table with masks, we can use the features built into `mast-table` to filter the data to fit our needs!

#### Opening the filters menu

In the upper left corner of our `mast-table` widget, we see a button labeled "Filters". Clicking this button (shown below) opens a menu where we can filter our table down to a smaller subset.


<img src="filters_button.png" width="150">


#### Applying filters

Let's start by filtering the data to only contain mid-infrared data. 

In the filters menu under "Add filter", we see a dropdown selection for the column to apply a filter to. By scrolling through the options or typing in the column name, select the column `instrume`. We can then select `MIRI` from the dropdown selection menu for "Value" below. Your filters menu should now look like this:

<img src="filter_menu.png" width="200">

Finally, we can hit "Apply Filter" and see the length of the table change as the the filter is applied. We now only have [MIRI (Mid-Infrared Instrument)](https://science.nasa.gov/mission/webb/mid-infrared-instrument-miri/) observations in our table. 

We also see the filter we just applied in the "Active filters" section. Click on the column name to expand the details and show exactly what filter is active!

The second filter we want is a small angular separation. 

Following the same process, select the column of interest `ang_sep`. Since the column contains only floats, we now see buttons to pick an operator and an input box with a slider to set the filter value. Select the operator, then use either the input box or the slider to restrict the data to those with an angular separation less than or equal to 0.3'.

Finally, hit "Apply Filter", and we again see the length of the table change as the the filter is applied.

We now see another filter appear in the "Active filters" section and can click on the column name to expand the details.

Your table should now look like this:

<img src="filtered_table.png" width="600">

#### Applying filters as match any or match all

Now, let's say we realized we don't want to exclude near-infrared observations of M1. We are now interested in data that is either mid-infrared data *or* has small angular separation.

In the upper right corner of our filters menu, we have a toggle `AND/OR` button, which controls if the active filters are applied as "match all" or "match any".

By toggling that to "or", we are now filtering our data to anything that is either mid-infrared data *or* has small angular separation!

#### Closing the filters menu

After applying these filters, we now want to be able to see more columns of the data, so we can again click on the "Filters" button to collapse the filters menu, and our `mast-table` now takes up the full width of the notebook!
