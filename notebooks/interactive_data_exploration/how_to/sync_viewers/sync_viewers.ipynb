<a id="what-is-a-viewport"></a>
### What is a Viewport?
A viewport is the part of an image currently visible in a viewer. Both `jdaviz` and `mast-aladin` implement the [Astro Image Display API](https://github.com/astropy/astro-image-display-api) (AIDA), which describes a viewport using properties such as its sky-coordinate center, field of view, rotation, and projection.

In this tutorial, we’ll sync the following viewport properties:
- **Center**: The center of the viewport in sky coordinates.
- **FOV**: The field of view of the viewport, measured along the viewer’s shorter side.
- **Rotation**: The rotation applied to the image in the viewport
- **Projection**: The current projection applied to the viewport

### Imports 
We will import the following in this tutorial:
- `mast-aladin`: to create sky viewers and manage synchronization.
- `jdaviz`: to create an image viewer.
- [`Sidecar`](https://github.com/jupyter-widgets/jupyterlab-sidecar): to arrange the viewers in the notebook.

```python
import jdaviz as jd

from mast_aladin import MastAladin, MastManager

from sidecar import Sidecar
```

### Widget Setup

<a id="initialize-widgets"></a>
**Initialize widgets:**
Before syncing viewers, we’ll create two `mast-aladin` viewers and one `jdaviz` viewer. We’ll point them at the Cartwheel Galaxy so their views can be compared.
- `mast_aladin`: loads the Cartwheel Galaxy.
- `mast_aladin_2mass`: loads the Cartwheel Galaxy using the 2MASS survey.
- `jdaviz`: loads a JWST NIRCam image of the Cartwheel Galaxy.

```python
mast_aladin = MastAladin(
    target="Cartwheel Galaxy",
    fov=1,
    height=500
)

mast_aladin_2mass = MastAladin(
    target="Cartwheel Galaxy",
    fov=1,
    height=500,
    survey='https://alaskybis.cds.unistra.fr/2MASS/Color/'
)

uri = 'mast:JWST/product/jw02727-o002_t062_nircam_clear-f090w_i2d.fits'
jdaviz = jd.App()
jdaviz.load(uri, format="Image", cache=True)
```

<a id="display-widgets"></a>
**Display widgets:** We’ll place each viewer in a `sidecar` so the views remain visible while you use the sync controls. Run the cell below to display the `jdaviz` app and both `mast-aladin` viewers. For more information on how this is done, please see the [how to organize widgets](../organize_widgets/organize_widgets.ipynb) tutorial.

```python
# Create a sidecar
jdaviz_sc = Sidecar(title="jdaviz", anchor="split-right")
with jdaviz_sc:
    jdaviz.show()

# Create a sidecar with defined anchor
mast_aladin_sc = Sidecar(title="mast_aladin", ref=jdaviz_sc, anchor="split-bottom")

with mast_aladin_sc:
    display(mast_aladin)

mast_aladin_2mass_sc = Sidecar(title="mast_aladin_2mass", ref=mast_aladin_sc, anchor="split-right")
with mast_aladin_2mass_sc:
    display(mast_aladin_2mass)
```

<a id="register-widgets"></a>
**Register widgets:** Next, we’ll register all three viewers with `MastManager`. This lets the `Viewer Sync UI` list them as possible sources and destinations.

```python
mm = MastManager()

mm.register_app(mast_aladin, "mast_aladin")
mm.register_app(mast_aladin_2mass, "mast_aladin_2mass")
mm.register_app(jdaviz, "jdaviz")
```

<a id="viewer-sync-ui"></a>
## Viewer Sync UI
The viewers are registered and visible. Run the cell below to open the Viewer Sync UI, then choose:

- **Source**: The widget whose viewport changes will be shared.
- **Destinations**: One or more widgets that receive those updates.
- **Aspects**: Center, Field of View, Rotation, and Projection.


```python
mm.display_plugin("viewer_sync")
```
