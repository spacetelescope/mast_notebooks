```python
from sidecar import Sidecar
from mast_aladin import MastAladin
import jdaviz as jd

from ipywidgets import widgets
```

## Organizing Widgets with `motorcycle`

### Setup
We will create 3 different sidecar examples, two with viz and one with generic HTML, to show off the different layouts that can be achieved with `motorcycle`.

```python
# Create a mast-aladin instance
mast_aladin = MastAladin()

# Create a sidecar with defined anchor
sc = Sidecar(title="mast-aladin Main", anchor="split-right")

# Display the widget within the sidecar
with sc:
    display(mast_aladin)
```

```python
# Create a sidecar with defined anchor
sc = Sidecar(title="jdaviz Main", anchor="split-bottom", ref=sc)

# Display a jdaviz instance widget within the sidecar
with sc:
    jd.show()
```

```python
# Create a generic sidecar widget
sc = Sidecar(title='Generic Sidecar Widget', anchor='split-right', ref=sc)

with sc:
    display(
        widgets.HTML(
            "<h2>generic sidecar widget</h2>"
        )
    )
```

### Layout Controls 
The controls for `motorcycle` are found in the "Layout" menu option at the end of the JupyterLab menu bar (after File, Edit, ...). `motorcycle` supports the following layout options:

- **All Tabs**: These commands will move all widgets, including `sidecars`, viz apps, and notebook tabs. 
- **Sidecars**: These commands will move all `sidecars`, including the viz apps and any others that are open in the workspace.
- **Viz apps**: These commands will only move instances of `jdaviz`, `mast-aladin`, or [`mast-table`](https://github.com/spacetelescope/mast-table).
