## Introduction

This tutorial describes how to load for non-Roman source catalogs into `mast-aladin`, allowing you to vizualize the data against a sky background. We will use the [astoryquery.mast](https://astroquery.readthedocs.io/en/stable/mast/mast.html) library to query for source catalogs from the [PanSTARRS](https://www2.ifa.hawaii.edu/research/Pan-STARRS.shtml) and [JWST](https://science.nasa.gov/mission/webb/) missions. The examples provided will focus on sources near the [Andromeda Galaxy (M31)](https://science.nasa.gov/mission/hubble/science/explore-the-night-sky/hubble-messier-catalog/messier-31/).


## Imports

We will import the following in this tutorial:

- astroquery.mast: query and download source catalogs for a subset of the astronomical catalogs stored at MAST
- mast_aladin.MastAladin: interactive sky map widget for visualizing source catalogs
- sidecar: organize our `mast-aladin` viewer widget

```python
from astroquery.mast import Catalogs, MastMissions
from mast_aladin import MastAladin
from sidecar import Sidecar
```

## Overlay non-Roman Source Catalogs

### Setup
The first step to working with `mast-aladin` is to setup the notebook workspace. We start by initializing a new instance of the `MastAladin` class with the `target` set to `M31` and the `fov` set to `10` to ensure we can see the Source Catalogs that will be loaded. The `Sidecar` library is used to display the viewer in a separate window. More information on this can be found in the [interactive_data_exploration/how_to/organize_widgets/organize_widgets.ipynb](../organize_widgets/organize_widgets.ipynb) tutorial.

```python
mast_aladin = MastAladin(target="M31", fov=10)

sc = Sidecar(title="mast-aladin", anchor="split-right")
# Display the widget within the sidecar
with sc:
    display(mast_aladin)
```

### Query for non-Roman Source Catalogs
The next step is to query for the non-Roman Source Catalogs. The [astroquery.mast.Catalogs](https://astroquery.readthedocs.io/en/stable/mast/mast_catalog.html) and [astroquery.mast.MastMissions](https://astroquery.readthedocs.io/en/latest/mast/mast_missions.html) classes provides access to a subset of the astronomical catalogs stored at MAST. For this example, we will look at the `PanSTARRS DR2` Source Catalogs and `JWST` Source Catalogs. The full list of Source Catalogs currently available through the classes are available in their respective documenation above. 

Both classes provide the same query API interface. The `query_object()` method returns an [Astropy Table](https://docs.astropy.org/en/stable/table/index.html) which is required for overlaying into `mast-aladin`.

**PanSTARRS Source Catalogs**

PanSTARRS has a large number of sources near M31, so the radius should be set to `.02` to ensure that the query can return without timing out.

```python
panstarrs_catalog = Catalogs.query_object("m31", radius=.02, catalog="PANSTARRS", data_release="dr2")
```

**JWST Source Catalogs**

JWST has no sources near the center of `M31`, so the radius needs to be set to `30` to ensure that results are returned.

```python
mm = MastMissions(mission="jwst")
jwst_catalog = mm.query_object("m31", radius=30)
```

### Overlaying a Source Catalog
You can overlay any `astropy.Table` with fields containing `ra` and `dec` values onto the sky viewer using the `add_table()` method. The method returns an `overlay` object which can be used to modify its styling properties.

```python
panstarrs_overlay = mast_aladin.add_table(panstarrs_catalog, name="panstarrs", raField="raMean", decField="decMean")
jwst_overlay = mast_aladin.add_table(jwst_catalog, name="jwst", raField="targ_ra", decField="targ_dec")
```

![alt text](overlaying-a-source-catalog.png)

### Update a Tables Syling Properties
You can update the name, color, shape, and source size of an overlay.

```python
# returns the updated Overlay
panstarrs_overlay = panstarrs_overlay.update(
    name="panstarrs-updated",
    color="Blue", 
    shape="circle", 
    source_size=30
)
```

![alt text](update-a-tables-styling-properties.png)

### Removing a Source Catalog
The `remove_overlay` method allows for the removal of Source Catalogs. 

First, identify the Source Catalogs to remove.

```python
mast_aladin.overlays
```

Next, provide the `name` attached to the Source catalog to the `remove_overlay` method.

```python
mast_aladin.remove_overlay('panstarrs-updated')
```
