### Introduction

In this tutorial, we’ll look at a source catalog from the _Roman Space Telescope_, which is a Level 4 (L4) product. Roman L4 catalogs are stored in <a href="https://parquet.apache.org/docs/file-format/">parquet format</a>. In standard Roman file naming conventions, files ending in `_cat.parquet` are single-band source catalogs. Using ``mast-aladin``, we can visualize the source catalog as markers on the sky as well as overlay a simulated Roman image on the all-sky atlas. Finally, we'll use ``mast-table`` to interactively inspect and filter the source catalog to select sources of interest.

### Imports 
We will import the following in this tutorial:
- `mast-aladin`: query and download source catalogs for a subset of the astronomical catalogs stored at MAST
- `mast-table`: interactively visualize the parquet catalog as a table
- `astropy.coordinates.SkyCoord`: define the coordinates of a target
- `astropy.table.Table`: read the parquet file
- (Optional) `astroquery.mast`: fetch Roman images and catalogs upon data release

```python
from mast_aladin import MastAladin
from mast_table import MastTable
from astropy.coordinates import SkyCoord
from astropy.table import Table
```

### Visualizing parquet files with ``mast-aladin``

### Initializing and displaying ``mast-aladin``

We define a SkyCoord instance for the target on which we want to center the HiPS image. Then we initialize `mast-aladin` with the background HIPS image set to 2MASS (colored) and centered on `target`. By default, calling `mast_aladin.show(title='mast-aladin-2mass')` launches a `sidecar` widget aligned to the right of your screen. 

```python
# create mast_aladin instance with 2MASS background
target = SkyCoord(ra=270, dec=66, unit="deg")
mast_aladin = MastAladin(
    survey='https://alaskybis.cds.unistra.fr/2MASS/Color/', # Accepts a HIPS survey url
    target=target,                  # accepts a string or SkyCoord instance
    fov=5,                          # field of view in degrees
)
mast_aladin.show(title='mast-aladin-2mass')
```

Using the `add_asdf` method, we can also display a Roman image overlaid on the 2MASS image.

```python
mast_aladin.add_asdf('data/r00004_r0_p01001_270p65x69y50_f146_coadd.asdf')
```

### Open and display a parquet catalog

Finally, we open the Roman parquet file with `astropy.table`. Using the `add_table` method, we add the source catalog.

```python
roman_catalog = 'data/r00004_r0_p01001_270p65x69y50_f146_cat.parquet'

source_catalog = Table.read(roman_catalog)

catalog_layer = mast_aladin.add_table(
        source_catalog, 
        shape='circle', 
        color='lime', 
        sourceSize=10,
        name='Roman_Catalog')
```

### Create a ``mast-table``

You can visualize and interact with the parquet catalog by using `mast-table`. The next cell initializes a `mast-table` widget, which you can use to filter, sort, and select interesting data.

```python
my_table = MastTable(source_catalog)
my_table
```
