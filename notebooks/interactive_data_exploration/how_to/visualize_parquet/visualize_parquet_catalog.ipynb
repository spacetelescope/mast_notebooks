# How to View Parquet Source Catalogs with `mast-aladin`
---

## Learning Goals 
This is a beginner tutorial on visualizing sources from a Roman Level 4 (L4) catalog using [`mast-aladin`](https://github.com/spacetelescope/mast-aladin). By the end of this tutorial, you will:
- Understand how to load an L4 parquet catalog into `mast-aladin`.
- Be able to visualize a source catalog overlaid on an image in `mast-aladin`.

In this tutorial, we’ll look at a source catalog from the _Roman Space Telescope_, which is a Level 4 (L4) product. Roman L4 catalogs are stored in <a href="https://parquet.apache.org/docs/file-format/">parquet format</a>. In standard Roman file naming conventions, files ending in `_cat.parquet` are single-band source catalogs.

### Imports 
We will import the following in this tutorial:
- `mast-aladin`: to create sky viewers and manage synchronization.
- `mast-table`: to interactively visualize the parquet catalog as a table.
- `astropy.coordinates.SkyCoord`: to define the coordinates of a target.
- `astropy.table.Table`: to read the parquet file. 
- (Optional) `astroquery`: to fetch Roman images and catalogs (but there are sample files available on Box:...)

````python
from mast_aladin import MastAladin
from mast_table import MastTable
from astropy.coordinates import SkyCoord
from astropy.table import Table
````



<div class="alert alert-block alert-info">
  Once Roman data are publicly available in the archive, you can search for and download L4 products like this:
    
```
import astropy.units as u
from astropy.coordinates import SkyCoord
from astroquery.mast import MastMissions

target = SkyCoord(ra=270, dec=66, unit="deg")
mast = MastMissions(mission='Roman')
roman_obs = mast.query_region(
     target,
     radius=0.3 * u.deg,
     optical_element="F146",
     select_cols="*",
 )
results = mast.get_unique_product_list(roman_obs)
mast.download_products(results[results['category']=='4'])['Local Path']
```

Until then, we load representative Roman query results from cache:
</div>

````python
roman_catalog = 'r00004_r0_p01001_270p65x69y50_f146_cat.parquet'
````

**Initialize a `mast-aladin` instance:** Next, we define a SkyCoord instance for the target on which we want to center the HiPS image. Then we initialize `mast-aladin` with the background HIPS image set to 2MASS (colored) and centered on `target`, with `fov` = 5 degrees and `height` = 500 pixels. Calling `mast_aladin.show(loc='sidecar:right')` launches a `sidecar` widget aligned to the right of your screen, but you can also create viewers `inline` or in `popout` mode.

````python
# initialize mast_aladin with 2mass background
target = SkyCoord(ra=270, dec=66, unit="deg")
mast_aladin = MastAladin(
    survey='https://alaskybis.cds.unistra.fr/2MASS/Color/', # Accepts a HIPS survey url
    target=target,                  # accepts a string or SkyCoord instance
    fov=5,                          # field of view in degrees
    height=500                      # widget height in pixels
)
mast_aladin.show(loc='sidecar:right', title='mast-aladin-2mass')
````

Using the `add_asdf` method, we can also display a Roman image overlaid on the 2MASS image.

````python
mast_aladin.add_asdf('r00004_r0_p01001_270p65x69y50_f146/r00004_r0_p01001_270p65x69y50_f146_coadd.asdf')
````

**Open and display a parquet catalog:** Finally, we open the Roman parquet file with `astropy.table`. Using the `add_table` method, we add the source catalog and set some optional aesthetic keyword arguments. These can be changed after the fact by clicking on the 'Stack' icon in `mast-aladin` and then the gear icon next to the "my parquet catalog" label.

````python
source_catalog = Table.read(roman_catalog)

catalog_layer = mast_aladin.add_table(
        source_catalog, 
        shape='circle', 
        color='lime', 
        sourceSize=10,
        name='my parquet catalog')
````

**Bonus:** You can interact with the parquet catalog using `mast-table`. The next cell initializes a `mast-table` widget. In the widget UI, you can use `Filters` to define logic that will filter your catalog, or you can simply select rows of interest.

````python
my_table = MastTable(source_catalog)
my_table
````
